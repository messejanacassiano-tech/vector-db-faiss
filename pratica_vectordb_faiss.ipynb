## Documento base e divisão em chuncks (com fonte citada)

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Texto conceitual sobre IA
# Fonte: Artigo Educacional sobre Fundamentos de IA / IFRN
documento_texto = """
A Inteligência Artificial (IA) é um campopda ciência da computação dedicado ao desenvolvimento de sistemas capazes de realizar tarefas que
normalmente exigiriam inteligência humana.
Entre suas subáreas, destacam-se o Aprendizado de Máquina (Machine Learning) e o processo de linguagem natural.
Os embeddings são representações vetoriais da palavras ou textos que capturam o significado semântico, permitindo que os modelos processem linguagem natural de
forma eficiente.
O Faiss (Facebook AI Similarity Search) é uma biblioteca desenvolvida para busca eficiente de similaridade e agrupamento de vetores densos em bancos de dados
vetoriais.
"""

# Divisão de documetno em chuncks conforme a prática da aula
text_splitters = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=20)
chunks = text_splitters.split_text(documento_texto)

print(f"Total de chuncks criados: {len(chunks)}")
```

## Embeddings e Banco Vetorial FAISS

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# Inicialização do modelo de embeddings e criação do VectorDB
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vector_db = FAISS.from_texts(texts=chunks, embedding=embeddings)

print("Banco de dados vetorial FAISS criado com sucesso!")
```

## RetrievalQA (Recuperação de Informação)

```python
# Pergunta de teste do usuário
# pergunta = "O que são embeddings e para que serve o FAISS?"
pergunta = "O que é Inteligência Artificial?"

#Recuperação dos trechos mais relevantes no VectorDB
docs_relevantes = vector_db.similarity_search(pergunta, k=1)

print("--- PERGUNTA DO USUÁRIO ---")
print(pergunta)

print("\n--- RESPOSTA RECUPERADA DO VECTORDB (FAISS) ---")
print(docs_relevantes[0].page_content)
```
